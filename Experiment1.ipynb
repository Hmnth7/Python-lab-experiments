```python
class User:
    def __init__(self, user_id, name, phone):
        self.user_id = user_id
        self.name = name
        self.phone = phone

    def display_details(self):
        print("\n----- USER DETAILS -----")
        print("User ID :", self.user_id)
        print("Name    :", self.name)
        print("Phone   :", self.phone)



class Customer(User):
    def __init__(self, user_id, name, phone, address):
        super().__init__(user_id, name, phone)
        self.address = address

    def place_order(self):
        print(f"\n{self.name} has placed an order successfully.")



class PremiumCustomer(Customer):
    def __init__(self, user_id, name, phone, address):
        super().__init__(user_id, name, phone, address)

    def calculate_discount(self, amount):
        discount = amount * 0.30  # 30% Discount
        print("\n----- DISCOUNT DETAILS -----")
        print("Original Amount :", amount)
        print("Discount (30%)  :", discount)
        print("Final Amount    :", amount - discount)
        return amount - discount


class Restaurant:
    def __init__(self, restaurant_id, name):
        self.restaurant_id = restaurant_id
        self.name = name

    def display_details(self):
        print("\n----- RESTAURANT DETAILS -----")
        print("Restaurant ID :", self.restaurant_id)
        print("Restaurant    :", self.name)


class DeliveryPartner:
    def __init__(self, partner_id, name):
        self.partner_id = partner_id
        self.name = name

    def assign_delivery(self):
        print(f"\nDelivery assigned to {self.name}.")


class Order:
    def __init__(self, order_id, item_name, quantity, price):
        self.order_id = order_id
        self.item_name = item_name
        self.quantity = quantity
        self.price = price

    def generate_bill(self):
        total = self.quantity * self.price
        print("\n----- BILL -----")
        print("Order ID :", self.order_id)
        print("Item     :", self.item_name)
        print("Quantity :", self.quantity)
        print("Price    :", self.price)
        print("Total    :", total)
        return total



customer = PremiumCustomer(
    101,
    "Hemanth kumar reddy",
    "9515462141",
    "Bangalore"
)

restaurant = Restaurant(
    201,
    "Pizza Hub"
)

delivery = DeliveryPartner(
    301,
    "Rahul"
)

order = Order(
    401,
    "Veg Pizza",
    2,
    350
)

# Display Details
customer.display_details()
restaurant.display_details()

# Place Order
customer.place_order()

# Generate Bill
bill = order.generate_bill()

# Calculate Discount
customer.calculate_discount(bill)

# Assign Delivery
delivery.assign_delivery()
```
