```python
class SavingsAccount(Bankaccount):
    def __init__(self, acc_no, name, balance=0):
        super().__init__(acc_no, name, balance)

    def calculate_interest(self):
        annual_rate = 0.01 # 1% annual interest
        interest = self._balance * annual_rate
        return interest
```

```python
class FixedDepositAccount(Bankaccount):
    def __init__(self, acc_no, name, balance=0, interest_rate=0.05, tenure_months=12):
        super().__init__(acc_no, name, balance)
        self.interest_rate = interest_rate
        self.tenure_months = tenure_months

    def calculate_interest(self):
        # Simple interest calculation
        interest = self._balance * self.interest_rate * (self.tenure_months / 12)
        return interest
```
